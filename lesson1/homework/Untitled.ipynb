```python
import pandas as pd 
from sqlalchemy import create_engine
from tqdm.auto import tqdm
import pyarrow.parquet as pq
import urllib.request
```

```python
prefix = "https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc"
url = f"{prefix}/taxi_zone_lookup.csv"
```

```python
df_iter = pd.read_csv(url, iterator=True, chunksize=10000)
```

```python
engine = create_engine("postgresql+psycopg://root:root@localhost:5432/homework_db")
```

```python
first = True
for df_chunk in tqdm(df_iter):
    if first:
        df_chunk.head(0).to_sql("zones", con=engine, if_exists="replace")
        first = False
    df_chunk.to_sql("zones", con=engine, if_exists="append")
print("-----Import Data Completely-----")
```
