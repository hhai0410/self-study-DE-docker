```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
prefix = 'https://d37ci6vzurychx.cloudfront.net/misc'
url = f'{prefix}/taxi_zone_lookup.csv'
```

```python
dtype = {
    "LocationID" : "Int64",
    "Borough" : "string",
    "Zone" : "string",
    "service_zone" :  "string"
}
df_iters = pd.read_csv(
    url,
    dtype = dtype,
    iterator = True,
    chunksize = 100)
```

```python
engine = create_engine("postgresql+psycopg://root:root@localhost:5432/ny_taxi")
from tqdm.auto import tqdm
```

```python
first = True
for df_chunk in tqdm(df_iters):
    if first:
        df_chunk.head(0).to_sql(name='zones', con = engine, if_exists = 'replace')
        first = False
    df_chunk.to_sql(name='zones', con = engine, if_exists = 'append')
```
